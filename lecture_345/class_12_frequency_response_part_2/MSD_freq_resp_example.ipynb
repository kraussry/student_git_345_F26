```python
import matplotlib.pyplot as plt
import numpy as np
from control import TransferFunction as TF
import control
```

```python
m = 1
b = 0.2
k = 30
```

```python
wn = np.sqrt(k/m)
b_cr = 2*np.sqrt(k*m)
z = b/b_cr
z
```

```python
wn
```

```python
G = TF(1/m,[1,2*z*wn,wn**2])
G
```

## Steady-State Fixed Sine Response vs. Full Response

```python
w = 2
```

```python
np.sqrt(-1+0j)
```

```python
s = 1j*w
s
```

```python
G(s)
```

```python
1/(s**2+0.2*s+30)
```

```python
C1 = np.real(G(s))
C1
```

```python
C2 = np.imag(G(s))
C2
```

```python
t = np.arange(0,30,0.01)
```

```python
y_ss = C1*np.sin(w*t) + C2*np.cos(w*t)
```

## Find Full Response with Transients

```python
u = np.sin(w*t)
```

```python
to, y_full = control.forced_response(G, t, u)
```

```python
plt.figure()
plt.plot(t,y_full)
plt.plot(t,y_ss,'--')
plt.xlabel('Time (sec.)')
plt.ylabel('$y(t)$')
plt.legend(['$y_{full}$','$y_{ss}$']) 
```
