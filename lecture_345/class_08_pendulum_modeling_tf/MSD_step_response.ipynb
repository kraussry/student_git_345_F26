```python
import matplotlib.pyplot as plt
import numpy as np
import control
from numpy import sin, cos, tan, pi
from control import TransferFunction as TF
```

```python
m = 2
k = 75
b = 1
```

```python
wn = np.sqrt(k/m)
wn
```

```python
b_cr = 2*np.sqrt(k*m)
b_cr
```

```python
z = b/b_cr
z
```

```python
a = z*wn
a
```

```python
C1 = 1/k
C1
```

```python
C3 = -C1
```

```python
C2 = (-z/k)/(np.sqrt(1-z**2))
C2
```

```python
t = np.arange(0,5,0.01)
```

```python
w = wn*np.sqrt(1-z**2)
```

```python
y_laplace = C1 + np.exp(-a*t)*(C2*np.sin(w*t) + C2*np.cos(w*t))
```

```python
G = TF(1/m,[1,2*z*wn,wn**2])
G
```

```python
to, y_step = control.step_response(G,t)
```

```python
plt.figure()
plt.plot(t,y_laplace, t, y_step,'--')
plt.xlabel('Time (sec.)')
plt.ylabel("Step Response")
```
