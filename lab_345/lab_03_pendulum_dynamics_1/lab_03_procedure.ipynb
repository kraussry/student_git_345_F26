# Pendulum Dynamics 1: Initial Condition and Impulse Response

## Overview

In this lab, you will investigate the accuracy of a pendulum equation of motion derived in class.  In the process, we will consider whether or not a pendulum can approximate a mass/spring/damper system.

You will  measure the response from two tests:

1. an initial displacement
2. impulse input

You will curve fit the data from those two tests to the expected response of a mass/spring/damper system.

In lecture, we derived the equation of motion for a pendulum to be 

$$\ddot{\theta} + \frac{3g}{2l} \theta = 0$$

using the small angle assumption (which probably means $\theta$ is less than 10$^\circ$).  This is the same as a mass/spring system with no damper where

$$\omega_n = \sqrt{\frac{3g}{2l}}$$

So, one of our questions will be whether or not this is a good estimate of for $\omega_n$.  

If you give the pendulum a small flick, you will note that the motion dies out.  So, the equation of motion for the pendulum might be

$$\ddot{\theta} + 2 \zeta \omega_n \dot{\theta} + \omega_n^2 \theta = 0$$

if the friction in the pendulum system can be approximated by viscous damping (spoiler alert, viscous damping is probably not a perfect model).

## Step 2: Collect Initial Condition Response Data

You will need to start the test in the Arduino serial monitor before releasing the pendulum. Give the pendulum an initial displacement for 5-10$\circ$ degrees and release it.  You should probably collect data for 10 seconds or so.  There are two ways to change how long a test runs:

1. You can change the value of `stop_t` in the Arduino template file
    - technically, you could just change it in the output `*.ino` file, but if you ever need to regenerate the code, your change will get overwritten
2. You could specify that `stop_t` is a menu parameter using the wxbd software.  This would mean that the Arduino would ask you over serial for the value of `stop_t` before each test. 

## Step 3: Curve Fit Initial Condition Response Data

Once you have the data from the Arduino serial monitor, copy and paste it into a text of csv file like in last week's lab.  Load the data into Python and use `optimize.fmin` to curve fit the data.  You can curve fit the data using a symbolic model based on solving the differential equation using Laplace or you could use a numeric approach inside your `mymodel` function.  

## Step 4: Collect Impulse Response Data

The impulse response test will be similar to the initial condition test performed in step 2, except that the pendulum will start at rest in the down position.  After starting the test in the Arduino serial monitor, you will gently whack the pendulum with a pen or pencil.  Note that flicking it with your finger is not recommended as it might hurt a little and it will probably be way too hard.

Copy and paste the input response data into a text or csv file as before.

## Step 5: Curve Fit Impulse Response Data

In order to help you get familiar with transfer function simulations using the Python `control` module, you are required to use `control.impulse_response` inside your `mymodel` function for this step.  The `optimize.fmin` varies the coefficients of your transfer function during the curve fitting process.  Since we do not have a way to measure how hard each team whacks the pendulum, you will need a coefficient that accounts for the unknown amplitude of the input.
