## Find the natural frequency of the pendulum

Using a stopwatch, flick the pendulum and measure how long it takes for a certain number of oscillations.  Estimate the natural frequency so that you know what counts as low and high frequency.

## Fixed Sine Tests

Once you the natural frequency of the pendulum, perform three fixed sine tests:

- low frequency (0.2-0.5 x the natural frequency)
- near resonance (approximately at the natural frequency)
- high frequency (3-5x the natural frequency)

You will need to guess and check a little to find a "good" input amplitude that overcomes friction without having the cart travel too far.

Save and plot the data for each test

- 3 different graphs (one for each frequency)
- input and output vs. time on each graph

### Fixed Sine Observations and Data Analysis

Record what you see qualitatively at the three different input frequencies.  Also, estimate the magnitude ratio and phase difference between the input and output for each frequency.

## Simulating Fixed Sine Response for an SDOF MSD System

Based on your experimental results, choose values for $\zeta$ and $\omega_n$ that approximately model your pendulum (perhaps you already have values from a previous lab).

Based on the standard form for a transfer function for an MSD system:

$$G(s) = \frac{1/m}{s^2 + 2 \zeta \omega_n s + \omega_n^2}$$

Simulate the response to the same three fixed sine inputs used in your experiments.  Use `control.forced_response` to perform the simulations.  

- How do the simulation results compare to the experimental ones?
- Find the magnitude ratio and phase for each simulation
    - how do the magnitude ratio and phase compare to the experimental results at each frequency?
