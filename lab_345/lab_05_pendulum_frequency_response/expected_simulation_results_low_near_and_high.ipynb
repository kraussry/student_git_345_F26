In my head, low frequency fixed sine response should mean that the input and output are mostly in phase and the magnitudes might be fairly close (but a scaling factor might be needed).

So, this is what I expected low the frequency simulation results to look like.  It turns out that the low frequency experiment didn't work out this way (I kind of expected some differences).
