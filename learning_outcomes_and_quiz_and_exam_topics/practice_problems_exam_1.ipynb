```python
def random_TF_mbk():
    k = rand()*200
    m = 5*rand()
    b_cr = 2*np.sqrt(k*m)
    z = 1.5*rand()
    b = z*b_cr
    G = TF(1,[m,b,k])
    print("m = %0.4g" % m)
    print("b = %0.4g" % b)
    print("k = %0.4g" % k)
    return G
```

```python
def random_TF_generator():
    wn = int(20*rand())
    z = rand()
    G = TF(1,[1,2*z*wn,wn**2])
    return G
```

```python
def check_zeta_wn(G):
    p_list = G.poles()
    p0 = p_list[0]
    wn = np.abs(p0)
    z = -np.real(p0)/wn
    return wn, z
```
