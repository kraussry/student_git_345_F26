This learning activities is designed to help you make the connection between frequency response at one frequency to Bode plots.  A Bode plot shows how magnitude and phase change over a wide range of frequencies.

## Problem Statement

Given the transfer function $G$ below:

- first find its steady-state response at a single frequency
- then use a `for` loop to repeat the analysis for 5-10 different frequencies
- then use another `for` loop to find magnitude $M$ and phase $\phi$ for 20-30 frequencies
    - plot these $M$ and $\phi$ values vs. frequency and overlay the results with a Bode plot for $G(j\omega)$

```python
z = 0.1
wn = 5*2*pi
```

```python
G = TF(wn**2, [1,2*z*wn,wn**2])
G
```

## Low Freqeuncy

```python
w_low = 1*2*pi
```

```python
s_low = 1j*w_low
```

```python
C1_low = np.real(G(s_low))
C2_low = np.imag(G(s_low))
```

```python
t_low_max = 3
```

```python
t_low = np.arange(0,t_low_max,0.01)
```

```python
y_low_ss = C1_low*np.sin(w_low*t_low) + C2_low*np.cos(w_low*t_low)
```

```python
u_low = np.sin(w_low*t_low)
```

```python
to, y_low_full = control.forced_response(G, t_low, u_low)
```

```python
plt.figure()
plt.plot(t_low,u_low, t_low, y_low_full,'--', t_low, y_low_ss,'-.')
```
