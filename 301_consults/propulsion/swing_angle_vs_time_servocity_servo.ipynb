```python
import matplotlib.pyplot as plt
import numpy as np
import control
from numpy import sin, cos, tan, pi
from control import TransferFunction as TF
```

```python
plt.rcParams['font.size'] = 14
```

```python
m = 45#kg approx 100lb
```

```python
g = 9.81
```

```python
l = 1.5#meters
```

```python
max_torque = 17#N*m based on 2450 oz*inches
```

```python
z = 0.1#over estimate of friction modeled with viscous damping
```

```python
wn = np.sqrt(g/l)
wn
```

```python
num = 1/(m*l**2)
num
```

```python
den = [1,2*z*wn,wn**2]
```

```python
G = TF(num,den)
```

```python
t = np.arange(0,20,0.01)
```

```python
tau = max_torque*np.sin(wn*t)
```

```python
to, theta_rad = control.forced_response(G,t,tau)
```

```python
theta_deg = theta_rad*180/pi
```

```python
plt.figure()
plt.plot(t,theta_deg)
plt.ylabel('$\\theta$ (deg.)')
plt.xlabel('Time (sec.)')
plt.tight_layout()
plt.savefig("swing_angle_vs_time.png",dpi=300)
```
